Print the total number of articles from the nested response dictionary.

```python
data = {
    "response": {
        "pages": 3,
        "total": 230
    },
    "search": {
        "keyword": "bitcoin",
        "page": 1
    }
}
```

```python
#Print the total number of articles from the nested response dictionary.
inner=data["response"]
print(inner["total"])
```

```python
#Change the number of pages in response to the integer 5, preserving other data.
inner["pages"]=5
print(data)
```

```python
#Change the search keyword to the string "gold".
search_key= data["search"]
search_key["keyword"]="gold"
print(data)
```

```python
#Increase the current search page by 1, using its existing value.
search_page=data["search"]
search_page["page"]=search_page["page"]+1
print(search_page)
```

```python
#Print the updated total page count, search keyword, and current search page, each on a separate line.
print(inner["pages"])
print(search_key["keyword"])
print(search_page["page"])
```
